# Question 1 — The question you could not answer  (15 points)

A tricycle operators' cooperative in Los Baños hires you. They hand you one file. It is called
`trips.xlsx` and it has one row for every trip a unit makes: the date, the unit number, the driver,
the route, and the total fare collected on that trip.

The manager asks you two things in the same breath:

> *"What is the average number of passengers per trip?"*
> *"And what is the average fare per passenger?"*

You look at the file for a long moment.

> *A photograph of a full jeepney tells you the jeepney was full.*
> *It does not tell you who was on it.*

### Answer all three parts

**(a)** State the grain of `trips.xlsx` as a sentence. Then explain which of the manager's two
questions the file can answer and which it cannot — using the difference between **data that is
missing** and **a limitation of the grain**. Say clearly which one this is.

**(b)** The manager says: *"Then just add a passenger column."* Explain, in business terms he would
accept, why that does not solve the second question — and what would actually have to change.

**(c)** You propose to rebuild. Name the **two tables** you would create, write the **grain of each**
as a full sentence, and name **one dimension both would share**. Say what that shared dimension lets
him do that he cannot do today.

### YOUR ANSWER — Question 1

*Write between the triple quotes. Do not delete them.*

```python
ANSWER_1 = """
(a) One row represents one trip per unit, per driver, per day, and per route, including the total fare per trip. It can answer the total fare per unit, but it cannot answer the average number of passengers per trip. The missing data is the number of passengers per trip. The limitation of this grain is that it cannot show the average because it captures only one row per trip.


(b) Adding another column would not be easy because it would require another table containing information about the passengers who ride the unit on each trip, as well as the fare for each trip based on the kilometers traveled by the unit. Missing data may be collected later. A grain limitation requires another table or a rebuild.


(c) The two tables I would create are a passenger table and a fare-rate-per-kilometer table for travel from point A to point B. For TblPassengers, the grain would be one row representing one person per trip. For tblFareRate, one row would represent one fare per distance traveled. The one dimension both tables share is the route or place. The shared dimension available today allows him to see which routes have more passengers and the places where the units most frequently travel and drop off passengers.
"""

print(f"Question 1 \u2014 {_wc(ANSWER_1)} words")
```

# Question 2 — The filter that only moved half the report  (15 points)

You have built a model for a hardware chain. One page shows **units sold** beside **units received**,
month by month, with a date slicer at the top.

A branch manager clicks **March**. The sales bars drop to March. The deliveries bars do not move at all.

Nothing turns red. No error appears. The page simply shows March sales against a whole year of
deliveries, and it looks like the branch is buried in stock it never sold.

Separately, you run a left anti join from the sales table to the branch list. **Forty rows come back.**

> *An auditor who finds ₱50,000 unaccounted for does not quietly adjust the ledger so it balances.*

### Answer all three parts

**(a)** List what must be true of a relationship for a filter to travel from a dimension to a fact —
name the **cardinality**, the **side** each table sits on, and what a marked date table is for.
Then name the specific fault behind the unmoving bars, and how you would prove it in one check.

**(b)** Explain what a **left anti join** is actually doing, in your own words. Then distinguish the
two kinds of orphan you might be looking at: one that is a **formatting fault** and one that is
**genuinely unknown**. Say how you would tell them apart, and how many of the forty you would expect
each kind to be — you may reason from your own model.

**(c)** You can repair the dimension, exclude the forty rows, or escalate and report without them.
**Choose one**, and state what the other two would have cost. Argue from this situation, not from a rule.

### YOUR ANSWER — Question 2

*Write between the triple quotes. Do not delete them.*

```python
ANSWER_2 = """
(a) For a dimension to connect to a fact, there should be no mismatch or problem with formatting, missing reference records, or invalid values. The key on which each table is based is BranchID, and a marked date table is used to view branch operations on a certain day. The specific fault behind the unmoving bars could be an incorrect type or format for the units sold or units received, which would violate validity. Using a left anti-join, we could identify unknown branches, incorrect date formatting, and incorrect data types in one check.


(b) A left anti-join looks at the rows in the first table that have no match in the second table; we can call the findings orphan keys. The two types of orphan keys we might identify are formatting errors involving the date and its data type, and genuinely unknown branches that have not yet been added to the branch list. I expect the errors to involve date-type formatting and mismatched BranchID values.


(c) I would report and escalate this to the manager so the dimension can be updated. If the list and data type are not updated, the units sold for March would not be recorded by branch. Essentially, removing this data would be like removing the business’s revenue.

"""

print(f"Question 2 \u2014 {_wc(ANSWER_2)} words")
```

# Question 3 — The model, a year from now  (15 points)

Your model works. The CFO likes it. You have been promoted to a different department and somebody
else now owns the file.

Opening it, they find:

- a table called `tblSuppliers` that no relationship touches
- a fact table whose grain you never wrote down anywhere
- a set of applied steps named `Changed Type`, `Changed Type1`, `Changed Type2`

> *You can eat a dish once. You can cook from a recipe forever.*
> *But a recipe with no measurements is just a list of ingredients.*

### Answer all three parts

**(a)** Explain **ETL** in your own words, then say what **ELT** changes and why it became popular.
Using that, state where the transformation lives in a Power Query model like yours — and what that
means for the person who inherits it.

**(b)** `tblSuppliers` joins to nothing. Explain why leaving it in the model is a **design fault**
rather than harmless clutter — name at least one specific thing that can go wrong because it is there.
Then say what you would do about it and what you would need to know first.

**(c)** Your grain statement currently exists only in your head. Name **two places** it could
physically live so that the next person finds it without asking you, and say what specifically goes
wrong when it lives nowhere. Use the naming of those applied steps in your answer.

### YOUR ANSWER — Question 3

*Write between the triple quotes. Do not delete them.*

```python
ANSWER_3 = """
(a) ETL first extracts the data from a source, transforms it, and then loads it into the warehouse or model. ELT changes the process by loading the data first and transforming it later in the warehouse or platform. This makes it popular because other people or successors can use the model once it is loaded in the cloud rather than stored only on your laptop. It is like a recipe that can be improved in the future. She is inheriting the process you created based on the lessons you learned from your previous mistakes. What she is not inheriting are the mistakes you made; she can improve upon the process.


(b) Even though tblSuppliers does not join to anything, it could be useful later if the CEO or manager wants to create a new table. Leaving it in the model is a design flaw rather than harmless clutter because it contains business information about the suppliers. One specific problem is that someone might try to use its key with the main table, but the key would point to nothing and add more columns. At this time, we may not need to use tblSuppliers, but we can leave it in the model because it could become useful later. For example, if the CEO wants to add supplier information for certain products by transaction, we could use it then.


(c) The two places where the grain statement could physically exist so that the next person can find it without asking me, allowing it to carry forward, are in the M statement and in the named applied steps. If it exists nowhere, the next person would not know about the previous mistakes and would inherit them.

"""

print(f"Question 3 \u2014 {_wc(ANSWER_3)} words")
```
