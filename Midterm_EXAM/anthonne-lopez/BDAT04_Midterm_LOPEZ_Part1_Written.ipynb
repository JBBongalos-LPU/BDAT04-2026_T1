```python
STUDENT_NAME = "Lopez, Val Anthonne C."        # <-- type your full name between the quotes
CASE_ASSIGNED = ""       # <-- P4, P5, P6 or P7

# ---------------------------------------------------------------- do not edit below
from datetime import datetime
try:
    from zoneinfo import ZoneInfo
    TZ = ZoneInfo("Asia/Manila")
except Exception:
    TZ = None

def _now():
    return datetime.now(TZ) if TZ else datetime.now()

START_TIME = _now()
ANSWER_1 = ANSWER_2 = ANSWER_3 = ""

def _wc(text):
    return len([w for w in str(text).split() if w.strip()])

def _surname(name):
    n = (name or "SURNAME").strip()
    if "," in n:                       # "Lagdamen, Lawrence M." -> Lagdamen
        return n.split(",")[0].strip().upper()
    parts = [p for p in n.replace("_", " ").split() if p.strip(".")]
    return (parts[-1] if parts else "SURNAME").upper()

def check_progress():
    """Word count for each answer, with a simple status."""
    rows = [("Question 1", ANSWER_1), ("Question 2", ANSWER_2), ("Question 3", ANSWER_3)]
    print(f"{'':<12}{'WORDS':>7}   STATUS")
    print("-" * 46)
    for label, text in rows:
        n = _wc(text)
        if n == 0:        status = "not started"
        elif n < 200:     status = "short \u2014 keep going"
        elif n <= 400:    status = "good length"
        else:             status = "long \u2014 that is allowed"
        print(f"{label:<12}{n:>7}   {status}")
    print("-" * 46)
    mins = int((_now() - START_TIME).total_seconds() // 60)
    print(f"Elapsed: {mins} minutes of 90")

print(f"Ready \u2014 {STUDENT_NAME or 'NAME NOT ENTERED'} ({CASE_ASSIGNED or '?'})")
print(f"Started at {START_TIME:%H:%M on %d %B %Y}  (Manila time)")
```
