# Question 1 — The question you could not answer  (15 points)

A tricycle operators' cooperative in Los Baños hires you. They hand you one file. It is called
`trips.xlsx` and it has one row for every trip a unit makes: the date, the unit number, the driver,
the route, and the total fare collected on that trip.

The manager asks you two things in the same breath:

> *"What is the average number of passengers per trip?"*
> *"And what is the average fare per passenger?"*

You look at the file for a long moment.

> *A photograph of a full jeepney tells you the jeepney was full.*
> *It does not tell you who was on it.*

### Answer all three parts

**(a)** State the grain of `trips.xlsx` as a sentence. Then explain which of the manager's two
questions the file can answer and which it cannot — using the difference between **data that is
missing** and **a limitation of the grain**. Say clearly which one this is.

**(b)** The manager says: *"Then just add a passenger column."* Explain, in business terms he would
accept, why that does not solve the second question — and what would actually have to change.

**(c)** You propose to rebuild. Name the **two tables** you would create, write the **grain of each**
as a full sentence, and name **one dimension both would share**. Say what that shared dimension lets
him do that he cannot do today.

### YOUR ANSWER — Question 1

*Write between the triple quotes. Do not delete them.*

```python
ANSWER_1 = """
(a) The trips.xlsx rows contains the logs of the trips of a particular date, route, bus, driver, conductor, depart time, fuel liters consumed, and odometer km. The manager asked the average number of passengers per trip and the average fare per passenger. The file provided cannot answer the manager's question because it shows the detail of the trip only, not the passengers'. The file can answer a lot of other things such as: what is the average fuel consumption? what is the total number of kilometers each vehicle has traveled? And it can not only answer questions about average. It could also answer questions like the daily toll fees, the number of trips a conductor has made in a certain span, etc. The data missing is the number of passengers per trip.


(b) For the first questions of the manager, if the data would contain the number of passengers per trip, that would mean that the data that will be collected is the total amount of money collected. It means that we can't really answer the question about the average fare per passenger. It would make sense if it was the average trip per day on a week or something like that. If we're going to do that, we would have to make a separate data for the individual records like the fare for that single passenger.


(c) I would keep the trips table, and add a new table called passengers. the trips table would contain the same columns, while the passengers table would contain the details about the specific trip, and fare of each passenger. The dimension they would both share is the TripID because it is needed for the passengers tab to be able to answer the question of the manager. That shared dimension can give him the information he need about the trip, but not his original question. This is because the data is not yet available. Once the data about the fare of each passenger is available, he would be able to get answers about the average fare per person.


"""

print(f"Question 1 \u2014 {_wc(ANSWER_1)} words")
```

# Question 2 — The filter that only moved half the report  (15 points)

You have built a model for a hardware chain. One page shows **units sold** beside **units received**,
month by month, with a date slicer at the top.

A branch manager clicks **March**. The sales bars drop to March. The deliveries bars do not move at all.

Nothing turns red. No error appears. The page simply shows March sales against a whole year of
deliveries, and it looks like the branch is buried in stock it never sold.

Separately, you run a left anti join from the sales table to the branch list. **Forty rows come back.**

> *An auditor who finds ₱50,000 unaccounted for does not quietly adjust the ledger so it balances.*

### Answer all three parts

**(a)** List what must be true of a relationship for a filter to travel from a dimension to a fact —
name the **cardinality**, the **side** each table sits on, and what a marked date table is for.
Then name the specific fault behind the unmoving bars, and how you would prove it in one check.

**(b)** Explain what a **left anti join** is actually doing, in your own words. Then distinguish the
two kinds of orphan you might be looking at: one that is a **formatting fault** and one that is
**genuinely unknown**. Say how you would tell them apart, and how many of the forty you would expect
each kind to be — you may reason from your own model.

**(c)** You can repair the dimension, exclude the forty rows, or escalate and report without them.
**Choose one**, and state what the other two would have cost. Argue from this situation, not from a rule.

### YOUR ANSWER — Question 2

*Write between the triple quotes. Do not delete them.*

```python
ANSWER_2 = """
(a) For a filter to travel from a dimension to a fact, the relationship should be one to many, where the dimension table is on one side and the fact table is on the many side. The relationship should also be active and allow the filter to move from the dimension to the fact table. A marked date table tells Power BI which table is being used as the main date table, which helps with date and time analysis.


(b) In my own words, a left anti join returns the rows from the left table that do not have a matching record in the right table. Based on the given, the formatting fault i've noticed is the 40 rows that come back. It could be a formatting problem like the capitalization, space, spelling, etc. In order to determine if this is a formatting fault or genuinely unknown, I would neeed to clean the values of the datas then check if anything comes back again. If no rows come back, it was a formatting problem. And if 40 rows still come back, it was genuinely unknown.


(c) I would repair the branch infomation first. I would not delete these rows because they might still be an important data about a sale. If I could not repair, I would report it back to the manager of the hardware chain and tell them that a branch cannot be confirmed.


"""

print(f"Question 2 \u2014 {_wc(ANSWER_2)} words")
```

Your model works. The CFO likes it. You have been promoted to a different department and somebody

**(c)** Your grain statement currently exists only in your head. Name **two places** it could
physically live so that the next person finds it without asking you, and say what specifically goes
wrong when it lives nowhere. Use the naming of those applied steps in your answer.
