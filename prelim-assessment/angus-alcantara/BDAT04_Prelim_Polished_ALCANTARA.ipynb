# Question 1 — The rows you did not keep

A new analyst joins a retail chain in Calamba. She is told to *"clean up the customer sales file before month-end."* She removes every row with an empty field, saves it, and sends it on.

The branch revenue report comes out **₱2.1 million lower** than the previous month. Nothing was wrong with the business.

> *A hospital chart with no allergies recorded is not a patient with no allergies.*
> *One means nobody asked. The other means somebody checked.*
> *Treating the first as the second is how people get hurt.*

### Answer all three parts

**(a)** Explain what has gone wrong, using the difference between a value that is **missing** and a value that was **never applicable**. Give one example of each from data you have worked with.

**(b)** Deleting rows is sometimes the correct decision. Describe a situation where you would be entitled to delete, and state **the test you would apply** before doing it.

**(c)** You delete 1,200 rows for a good reason. What do you owe the person who reads your report — and **where, exactly, does that obligation get recorded?**


### YOUR ANSWER — Question 1

*Write between the triple quotes. Do not delete them.*


```python
ANSWER_1 = """
(a) She removed every row that had an empty field. If the empty field was an email field, all rows with blank email columns would be removed, and a percentage of the branch revenue would be lost. Removing anything blindly, without knowing what you are removing, is faulty because those rows are part of the business facts. One example is when 45% of the rows are null, not because of a data defect, but because of how the business operates: customers do not need to apply for an account or provide information.

(b) Deleting rows is sometimes the correct decision, and I would be entitled to delete a row if it were a duplicate because it would artificially inflate the company’s revenue. The test I would apply is to use Keep Duplicates in Power BI and record the number of rows, then remove that step and proceed to remove rows by removing duplicates.

(c) I owe it to the manager to tell him about the excess sales transactions that were entered twice and could affect the expected revenue. This can also be seen in the future by anyone who works for the company, as it is recorded in the Applied Steps with an M comment and can be seen in the TMDL view.
"""

print(f"Question 1 — {_wc(ANSWER_1)} words")
```

# Question 2 — The number you could not explain

It is Thursday. The Operations Manager needs branch figures on Friday morning for a meeting with the owner.

You find two things:

- **Forty sales** are recorded against branches that do not exist in the branch list.
- **Nine sales** are for a product that is not in the product catalogue.

You do not know why either is true, and you cannot find out before Friday.

> *An auditor who finds ₱50,000 unaccounted for does not quietly adjust the ledger so it balances.*
> *A balanced ledger is not the goal. A true one is.*

### Answer all three parts

**(a)** Explain what these two findings tell you about the data, using the correct terms. **Which quality dimension does each one fail?**

**(b)** Write what you would actually **say to the Operations Manager on Friday** — three or four sentences, as you would say them. Then state **one thing you would refuse to do**, and why.

**(c)** Two ways this can go wrong. You could raise a problem that turns out to be nothing. Or you could stay quiet about one that turns out to matter.

   **Which error costs more here, and why?** Your answer should depend on the situation, not on a rule.


### YOUR ANSWER — Question 2

*Write between the triple quotes. Do not delete them.*


```python
ANSWER_2 = """
(a) These two findings tell me that the data lacks timeliness, so the directory needs to be updated because the data should be current enough to support the decisions being made. They also indicate a lack of accuracy because the directory is not accurate enough; the values it reflects are not the facts it claims to represent.

(b) What I would tell the Operations Manager is to do his job, as he is not fulfilling his duty to provide timely and accurate data. If he continues working the way he does, how can I help the business make better decisions based on data that is not accurate or up to date? I need to trust the data before we can trust the decisions we build from it. The one thing I would refuse to do is update the directory because that is not my job. I would also refuse to remove the forty sales from a branch that does not exist in the master list or the nine sales for a product that is also not in the master list because, again, it is not my job. It is the Operations Manager’s job, and I will not guess or edit the raw data.

(c) Staying quiet about something that turns out to matter would cost more because it would be negligence on my part if those transactions were actually valid and I invalidated them because of the untimeliness of the data caused by the Operations Manager. Therefore, I would raise the problem with him, and it could turn out to be nothing. If so, that is fine because I fact-checked it. However, if I stayed quiet and it turned out to matter, it could lead to incorrect business decisions. Not using the forty sales from the unknown branch and the nine sales of the unknown products would make the data unreliable and remove a certain amount from the business’s total revenue.
"""

print(f"Question 2 — {_wc(ANSWER_2)} words")
```

# Question 3 — Monday morning, without you

You resign. Your last day is Friday.

On Monday a new analyst sits at your desk and opens your Power BI file. The CFO needs the profitability report by Tuesday. Some of the underlying data is **employee records** — names, daily rates, employment status.

You are not reachable.

> *You can eat a dish once. You can cook from a recipe forever.*
> *The question is which one you left behind.*

### Answer all three parts

**(a)** Explain the **ETL process** in your own words, and use it to say what your successor is actually inheriting — and what she is **not**.

**(b)** She clicks Refresh and it fails. **List what must have been true of your file** for that click to have worked, and explain why ***refreshable*** is a higher standard than ***correct***.

**(c)** Three obligations do not end when you leave the building:

   - the **privacy** of the people in that employee data
   - an **honest record** of what you changed and why
   - a **declaration of any AI** you used

   Take **any two** and explain what specifically you would have had to do, **before Friday**, for each one to survive your departure.


### YOUR ANSWER — Question 3

*Write between the triple quotes. Do not delete them.*


```python
ANSWER_3 = """
(a) The ETL process is the data pipeline from Power BI into the report. First, you export the data from Power BI or the cloud so it is not stored only on your laptop and can be used by your successor. Then, you transform the data through the steps you made during the data warehousing process. Each step should have an understandable title and an M comment so your successor can understand why you made it. Finally, you load the data so it can be useful to the person who inherits it. It is like a recipe that can be improved in the future. She is inheriting the process you created based on the lessons you learned from your previous mistakes. What she is not inheriting are the mistakes you made; she can improve upon the process.

(b) If she clicks Refresh and it fails, then the source path of my file has not been properly configured. For it to work, I have to parameterize the path so that the model works on every machine and can be refreshed anywhere. It is like centralizing the source path in one folder, from which the successor can change the source path. Being refreshable means that the model works across different machines, while a file that works on only one machine does not take the future into consideration.

(c) The number one priority is the privacy of the employees because the employee data contains sensitive information, and we are subject to the strict requirements of RA 10173. It should be treated with full responsibility and must not be leaked or fed into LLMs. I would also keep an honest record of what I changed and why, as it is the recipe I created. I would write it in a notebook or an MD file so the successor can understand what I did. I would also use clear names for the steps and include an M comment for every step that was created so the process can survive my departure.
"""

print(f"Question 3 — {_wc(ANSWER_3)} words")
```
