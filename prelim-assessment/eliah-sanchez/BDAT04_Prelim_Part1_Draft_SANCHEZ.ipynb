# Question 1 — The rows you did not keep

A new analyst joins a retail chain in Calamba. She is told to *"clean up the customer sales file before month-end."* She removes every row with an empty field, saves it, and sends it on.

The branch revenue report comes out **₱2.1 million lower** than the previous month. Nothing was wrong with the business.

> *A hospital chart with no allergies recorded is not a patient with no allergies.*
> *One means nobody asked. The other means somebody checked.*
> *Treating the first as the second is how people get hurt.*

### Answer all three parts

**(a)** Explain what has gone wrong, using the difference between a value that is **missing** and a value that was **never applicable**. Give one example of each from data you have worked with.

**(b)** Deleting rows is sometimes the correct decision. Describe a situation where you would be entitled to delete, and state **the test you would apply** before doing it.

**(c)** You delete 1,200 rows for a good reason. What do you owe the person who reads your report — and **where, exactly, does that obligation get recorded?**


### YOUR ANSWER — Question 1

*Write between the triple quotes. Do not delete them.*


```python
ANSWER_1 = """
(a) The new data analyst might have deleted rows that would be useful in reporting the branch revenue for the whole month. From the given analogy "A hospital chart with no allergies recorded is not a patient with no allergies". That said, the record would still be useful. It only indicates that the patients' data has no allergies recorded but it can still be used to record other information like their names, hospital bill, etc. If we apply this to the work of the new data analyst, that would mean a row with empty fields does not mean that you need to delete these rows as it may contain information about product sales to make the revenue report accurate.


(b) There are several situations where you are entitled to delete rows: if the row does not provide an information that could be useful for the report, a customer returns their purchased item, or if you are simply instructed to delete specific rows with certain something. The test that I would apply is by checking the total revenue and at the removing the empty fields to check if there would be a difference in the total revenue.


(c) I owe the person who reads my report the summary and the details of the rows i deleted. A full report on what I deleted and the exact reason behind it. This is so that the person that I'm passing my work to be able to get the information they need without rows that would be no use to them. That obligation could get recorded through a log in the form of either an md file or a pbix file.

"""

print(f"Question 1 — {_wc(ANSWER_1)} words")
```

# Question 2 — The number you could not explain

> *An auditor who finds ₱50,000 unaccounted for does not quietly adjust the ledger so it balances.*
> *A balanced ledger is not the goal. A true one is.*

### Answer all three parts

**(a)** Explain what these two findings tell you about the data, using the correct terms. **Which quality dimension does each one fail?**

**(b)** Write what you would actually **say to the Operations Manager on Friday** — three or four sentences, as you would say them. Then state **one thing you would refuse to do**, and why.

**(c)** Two ways this can go wrong. You could raise a problem that turns out to be nothing. Or you could stay quiet about one that turns out to matter.

   **Which error costs more here, and why?** Your answer should depend on the situation, not on a rule.


### YOUR ANSWER — Question 2

*Write between the triple quotes. Do not delete them.*


```python
ANSWER_2 = """
(a) By having forty sales that are recorded with branches that the Operations Manager require a record of, it tells me that the data is complete because there are records from other branches. But on the other hand, having nine sales of a product that is not in the product catalogue could mean that the data the employees recorded was a manually inputted record of a sale. The quality dimension that the forty sales recorded is Completeness. This is because there are unnecessary values needed in the data. For the nine sales of a product that is not in the catalog, it would be Consistency. It lacks consistency because all products must have a category in the product catalogue.


(b) I would tell the operations manager my findings from the data provided was inconsistent. Having inconsistent data would make the report of the branch figures incomplete like the nine products that are not in the catalogue. It doesn't matter how little the count is for data that is incomplete or inconsistent. One thing I would refuse to do is giving them the exact branch figures. I would ask for a data that is cleaned and complete.


(c) A problem that I could raise that would be nothing is the forty sales recorded against branches that are not in the list. I could just delete or ignore that data. If I stay quiet about the problem of the nine products, that would mean that the report for specific product would be incomplete. This means that this would be the error that is most costly because it would lack the consistency and completeness when reporting for example: Which product is the best selling in this specific branch?

"""

print(f"Question 2 — {_wc(ANSWER_2)} words")
```
