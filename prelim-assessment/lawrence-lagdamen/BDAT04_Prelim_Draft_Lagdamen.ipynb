# Question 1 — The rows you did not keep

A new analyst joins a retail chain in Calamba. She is told to *"clean up the customer sales file before month-end."* She removes every row with an empty field, saves it, and sends it on.

The branch revenue report comes out **₱2.1 million lower** than the previous month. Nothing was wrong with the business.

> *A hospital chart with no allergies recorded is not a patient with no allergies.*
> *One means nobody asked. The other means somebody checked.*
> *Treating the first as the second is how people get hurt.*

### Answer all three parts

**(a)** Explain what has gone wrong, using the difference between a value that is **missing** and a value that was **never applicable**. Give one example of each from data you have worked with.

**(b)** Deleting rows is sometimes the correct decision. Describe a situation where you would be entitled to delete, and state **the test you would apply** before doing it.

**(c)** You delete 1,200 rows for a good reason. What do you owe the person who reads your report — and **where, exactly, does that obligation get recorded?**


### YOUR ANSWER — Question 1

*Write between the triple quotes. Do not delete them.*


```python
ANSWER_1 = """
(a)
In the case of the new analyst, she was given a set of data and was instructed to clean up the “customer sales file” before month-end. Upon analysing the file, she purposely deleted rows where “null” data is present which resulted in revenue report discrepancy. It was stated that after what the analyst decided on doing, the branch’s sales revenue resulted in 2.1 million worth of unaccounted sales data.

In the case of data analysing, a “null” reading doesn’t always mean that data was never present. For example, in the case of Kape Tayo where the company’s “tblSales [CustomerID]” resulted in a total of 4775 “null” reading. That data does not suggest that there were “4775 ghost customers” it simply suggests that those customers were never registered in the company’s system and are walk-in customers.

(b)
Deleting rows in a business data without performing further investigation would simply mean a lack of analytical mindset for a data analyst.

In the case of Kape Tayo data upon further investigation, duplicates data sets were found in their “tblSales [SaleID]” this would either suggest a problem in their system, or a manual data entry fault. A total of 46 duplicate rows were found, but that information alone does not suggest removing all the 46 duplicate rows. It simply suggest that these values were doubled hence, only 23 rows should be removed in the dataset.

(c)
As mentioned in my answer to “B.” deleting unnecessary data such as duplicates will make the data set more feasible in terms of business data analytics.

However, a good analyst should always report what is/was deleted and it should be seen in their final report.

"""

print(f"Question 1 — {_wc(ANSWER_1)} words")
```

# Question 2 — The number you could not explain

It is Thursday. The Operations Manager needs branch figures on Friday morning for a meeting with the owner.

You find two things:

- **Forty sales** are recorded against branches that do not exist in the branch list.
- **Nine sales** are for a product that is not in the product catalogue.

You do not know why either is true, and you cannot find out before Friday.

> *An auditor who finds ₱50,000 unaccounted for does not quietly adjust the ledger so it balances.*
> *A balanced ledger is not the goal. A true one is.*

### Answer all three parts

**(a)** Explain what these two findings tell you about the data, using the correct terms. **Which quality dimension does each one fail?**

**(b)** Write what you would actually **say to the Operations Manager on Friday** — three or four sentences, as you would say them. Then state **one thing you would refuse to do**, and why.

**(c)** Two ways this can go wrong. You could raise a problem that turns out to be nothing. Or you could stay quiet about one that turns out to matter.

   **Which error costs more here, and why?** Your answer should depend on the situation, not on a rule.


### YOUR ANSWER — Question 2

*Write between the triple quotes. Do not delete them.*


```python
ANSWER_2 = """
(a)
These two findings suggest either a “consistency” or “validity” error in the data set. The Forty sales that were recorded against branches that do not exist in the branch list would suggest that the BranchID has formatting issues resulting to these forty sales not having their respective BranchID. The same situation would be relevant with the Nine sales reported to not belong in any of the product catalogue.

However, this assumption alone does not suggest my theory to be true. A good analyst would have to re-analyze the data.

(b)
I would say, “I found 2 discrepancies in our branch data that needs more investigation, thus TIME to find out what really happened to it. I cannot simply present you a false data just to meet with the deadline. I would need to verify this data with other available branch data and would need to cross reference them to find out whether my theory about it being a formatting issue is true or not”

I would not simply report these discrepancies in the data and leave it at that, my first instinct would be to investigate the circumstance even more so that I could present a better and working data to you.

(c)
In my life, I go by the saying “Prevention is better than cure.” and I stand by this saying be it with my work, personal life, and in school.

In the case of the missing values in the data set, it is always better to investigate more rather than just make assumptions that would inevitably cause a problem in the future. Masking a problem does not always mean solving it.
"""

print(f"Question 2 — {_wc(ANSWER_2)} words")
```

# Question 3 — Monday morning, without you

You resign. Your last day is Friday.

On Monday a new analyst sits at your desk and opens your Power BI file. The CFO needs the profitability report by Tuesday. Some of the underlying data is **employee records** — names, daily rates, employment status.

You are not reachable.

> *You can eat a dish once. You can cook from a recipe forever.*
> *The question is which one you left behind.*

### Answer all three parts

**(a)** Explain the **ETL process** in your own words, and use it to say what your successor is actually inheriting — and what she is **not**.

**(b)** She clicks Refresh and it fails. **List what must have been true of your file** for that click to have worked, and explain why ***refreshable*** is a higher standard than ***correct***.

**(c)** Three obligations do not end when you leave the building:

   - the **privacy** of the people in that employee data
   - an **honest record** of what you changed and why
   - a **declaration of any AI** you used

   Take **any two** and explain what specifically you would have had to do, **before Friday**, for each one to survive your departure.


### YOUR ANSWER — Question 3

*Write between the triple quotes. Do not delete them.*


```python
ANSWER_3 = """
(a)
The ETL process simply means the the new analyst should be able to:
E - Extract the data
T - Transform the data
L - Load the data
My successor is actually inheriting the data that I have worked on during my time in the company. And being an analyst who just resigned and has the responsibility to have the datasets ready for the next person, my successor should be able to do all these steps without any problems.

(b)
Given the situation and based on my personal experience, one problem that may have caused the data to not load would be because the supplemental files were ran in my personal computer rather than on an open source environment where the data can be accessed anywhere and by anyone who needs to work on it.

Refreshable data is always the standard because a dataset should always retain its integrity even when the person who initially worked on it is no longer available or accessible.

(c)
The two things that I will work on before I leave by Friday, would be the honest record of what I changed and why, along with the AI declaration while I worked on the datasets provided to me by the company.

By doing this, my successor would have a clear path to take when dealing with the datasets in the future. All decisions are accounted for and explained thoroughly.
"""

print(f"Question 3 — {_wc(ANSWER_3)} words")
```
